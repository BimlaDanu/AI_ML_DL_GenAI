```python
## pip install openai
import os
from openai import OpenAI

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"],
)

response = client.chat.completions.create(
    model="meta-llama/llama-3.1-8b-instruct",
    messages=[{"role": "user", "content": "What is a large language model?"}],
)
print(response.choices[0].message.content)
```

```python
import os
from openai import OpenAI
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"],
)

# Define the parameters for the API call
response = client.chat.completions.create(
    model="openai/gpt-5-mini",  # Specify the model to use
    messages=[
        {"role": "system", "content": "You are a knowledgeable assistant."},
        {"role": "user", "content": "Explain how machine learning works in simple terms."}
    ],
    temperature=0.7,         # Adjust creativity (higher = more random)
    top_p=0.9,               # Nucleus sampling threshold
    max_tokens=150,          # Limit response length
    frequency_penalty=0.5,   # Penalize repetitive phrases
    presence_penalty=0.3,    # Encourage introducing new concepts
)
# Extract and print the model's response
print(response.choices[0].message)
#print(response.choices[0].message.content)
```
